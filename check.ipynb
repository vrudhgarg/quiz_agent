```python
import pymupdf

doc = pymupdf.open("Lecture_4.pdf")  # open a document

    
```

```python
from PIL import Image
import io
import pytesseract
pix = doc[0].get_pixmap()
img = Image.open(io.BytesIO(pix.tobytes("png")))
text = pytesseract.image_to_string(img)
```

```python
# Increase resolution (default is 72 dpi, try 300)
pix = doc[1].get_pixmap(matrix=pymupdf.Matrix(300/72, 300/72))
img = Image.open(io.BytesIO(pix.tobytes("png")))

text = pytesseract.image_to_string(img)
```
